## Ejercicio 1. área de un círculo

```python
import math

#La fórmula como expresión

radio=5
area=math.pi * radio**2
print(area)

#La misma fórmula dentro de una función que devuelve el valor

def area_circulo(radius):
    return math.pi*radius**2

#Prueba con los radios r=0; r=1; r=5

print(area_circulo(0))
print(area_circulo(1))
print(area_circulo(5))
```

## Ejercicio 2. Celsius a Fahrenheit

```python
# La fórmula como expresión

C=0
F=C*(9/5)+32
print(F)

#La misma fórmula dentro de una función que devuelve el valor

def farenheit(celsius):
    return celsius*(9/5)+32

#Prueba con diferentes temperaturas

print(farenheit(0))
print(farenheit(10) , "°F")
```

## Ejercicio 3. Promedio de tres números

```python
# La fórmula como expresión

x=2
y=3
z=4
prom=(x+y+z)/3
print(prom)

#La misma fórmula dentro de una función que devuelve el valor

def prom (x,y,z):
    return (x+y+z)/3

#Prueba con diferentes valores

print(prom(1,2,3))
print(prom(2,3,4))
```
